```python
# O método info() é uma ferramenta essencial para inspeção detalhada da estrutura de um DataFrame pandas. Ele fornece um resumo conciso mas completo 
# sobre o conjunto de dados:
# 1. RangeIndex: Índice do DataFrame -> Mostra o intervalo de índices (0 a N-1);
# 2. Non-Null Count: Contagem de valores não-nulos -> Indica quantos valores válidos existem em cada coluna;
# 3. Dtype: Tipo de dado:
#    3.1. object: Strings/texto;
#    3.2. int64: números inteiros;
#    3.3. float64: Números decimais;
#    3.4. bool: Valores booleanos;
#    3.5. datetime64: Datas.
```

```python
# O método describe() é uma ferramenta poderosa para análise estatística descritiva de DataFrames e Series pandas. Ele gera automaticamente estatísticas
# resumidas que ajudam a entender a distribuição e características dos dados.
# 1. count: Número de valores não-nulos;
# 2. mean: Média aritmética;
# 3. std: Desvio padrão;
# 4. min: Valor mínimo;
# 25%: Primeiro quartil (percentil 25);
# 50%: Mediana (segundo quartil);
# 75%: Terceiro quartil (percentil 75);
# max: Valor máximo.
```

```python
df.describe()
```

```python
# O método describe com o parâmetro (include="O") é uma variação do método describe() do pandas que se concentra especificamente em colunas com dados 
# textuais (objetos) em um DataFrame.
# Quando usado com o include="O", o método describe() fornece estatísticas descritivas diferentes das numéricas,específicas para colunas de texto:
# 1. count: Número de valores não-nulos;
# 2. unique: Quantidade de valores únicos/distintos;
# 3. top: Valor mais frequente (moda);
# 4. freq: Frequência do valor mais comum.
```

```python
df.describe(include="O")
```

```python
# O método duplicated() é usado para identificar linhas duplicadas em um DataFrame ou Series pandas. Ele retorna uma série booleana onde True indica que a
# linha é duplicada (aparece mais de uma vez) e False indica que a linha é única. Em conjunto com o método sum() informa o número total de linhas repetidas.
```

```python
df.duplicated().sum()
```

```python
# O método isnull() é usado para identificar valores ausentes ou nulos (missing values) em um DataFrame ou Series pandas. Ele retorna um objeto do mesmo 
# tamanho onde True indica que o valor é nulo e False indica que o valor é válido. Em conjunto com o método sum() mostra a contagem de nulos por coluna.
```

```python
df.isnull().sum()
```

```python
# Realizando uma filtragem que seleciona apenas as linhas onde a coluna 'Postal Code' contém valores nulos/ausentes. O resultado é um novo DataFrame 
# contendo apenas os registros problemáticos (sem código postal).
```

```python
df[df['Postal Code'].isnull()]
```

```python
# Uma forma de resolver o problema com valores ausentes é utilizando o método fillna() que preencherá os campos selecionados do DataFrame com o valor 
# especificado entre parênteses, que nesse caso será o valor 5401.
```

```python
df['Postal Code'] = df['Postal Code'].fillna(5401)
df
```

```python
# Conferindo com o método isnull().sum() que não existe mais nenhum valor nulo/ausente no DataFrame.
```

```python
df.isnull().sum()
```

```python
# O método set_index transforma a coluna Row ID no novo índice do DataFrame, substituindo o índice numérico padrão (0,1,2,...). O parâmetro inplace=True 
# faz alteração diretamente no DataFrame original, sem precisar criar uma cópia.
```

```python
df.set_index('Row ID', inplace=True)
df
```

```python
# O código a seguir calcula três estatísticas resumidas (mínimo, máximo e média) sobre os valores da coluna 'Sales' (Vendas) do DataFrame.
```

```python
df['Sales'].agg(["min", "max", "mean"])
```

```python
# O método value_counts() conta a frequência de cada valor único em uma coluna (Series) do DataFrame. Neste caso contará a frequência de valores únicos na
# coluna Category.
```

```python
df['Category'].value_counts()
```

```python
df['Sub-Category'].value_counts()
```

```python
# O método pivot_table cria uma tabela dinâmica que calcula a média das vendas (Sales) para cada categoria (Category) presente no DataFrame.
```

```python
category_mean=df.pivot_table(index="Category", values="Sales", aggfunc="mean")
category_mean
```

```python
# O método a seguir substitui um valor específico por outro em uma coluna do DataFrame. No caso, ele localiza todas as ocorrências do texto "United States"
# na coluna 'Country' e as substitui por "US".
```

```python
df['Country'] = df['Country'].replace("United States", "US")
```

```python
df
```

```python
# O método rename renomeia uma ou mais colunas do DataFrame. No exemplo a seguir, ele altera o nome da coluna 'Country' para 'The Country'. O parâmetro
# inplace=True faz a alteração diretamente no DataFrame original, sem precisar criar uma cópia.
```

```python
df.rename(columns={"Country":"The Country"}, inplace=True)
df
```

```python
# O código a seguir converte colunas de texto (ou outros tipos) para o formato de data (datetime) do pandas. Ele transforma as colunas 'Order Date' (Data 
# do pedido) e 'Ship Date' (Data do Envio) em datas propriamente ditas, permitindo operações temporais como calcular diferenças, extrair mês/ano, etc. 
```

```python
df['Order Date']=pd.to_datetime(df['Order Date'], dayfirst=True, errors="coerce")
df['Ship Date']=pd.to_datetime(df['Ship Date'], dayfirst=True, errors="coerce")
```

```python
df.info()
```

```python
# Este código cria uma nova coluna chamada 'shipping_days' que armazena a diferença em dias entre a data de envio (Ship Date) e a data do pedido 
# (Order Date). O resultado é um objeto Timedelta (diferença temporal) que representa o intervalo entre as duas datas.
```

```python
df['shipping_days']=df['Ship Date'] - df['Order Date']
df
```

```python
# Este código encontra o maior valor (máximo) na coluna 'shipping_days' do DataFrame e armazena esse valor na variável max_days. É uma forma rápida de 
# identificar o tempo máximo que algum pedido levou para ser enviado.
```

```python
max_days=df['shipping_days'].max()
max_days
```

```python
# Este código encontra todas as cidades onde ocorreu o maior tempo de envio (shipping_days igual ao valor máximo) e retorna uma lista com os nomes únicos
# dessas cidades.
```

```python
city_of_max = df[df['shipping_days'] == max_days]['City'].unique()
city_of_max
```

```python
# Listando as cidades com maior valor encontrado de shipping_days
```

```python
for city in city_of_max:
    print(city)
```
