### Exercício 1: Gráfico de Uma Função Linear

Crie um gráfico para a função y = 2x + 3 para valores de x no intervalo de 5 a 10.

```python
import matplotlib.pyplot as plt

def linear():
    lista_x = []
    lista_y = []
    for x in range(5, 11):
        lista_x.append(x)
        y = 2 * x + 3     
        lista_y.append(y)
    print(lista_x)    
    print(lista_y)

    plt.plot(lista_x, lista_y, marker='o', linestyle='-', color='b')
    plt.title('Gráfico de uma Função Linear')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.grid(True)
    plt.show()

linear()

```

### Exercício 2: Gráfico de uma Função Quadrática

Desenhe o gráfico da função y = x^2 - 4x + 4. Crie uma massa de dados de exemplo usando np.linspace().

```python
import numpy as np
import matplotlib.pyplot as plt

def quadratica():
    dados = np.linspace(0, 10, 50)
    lista_x = []
    lista_y = []
    for x in dados:
        lista_x.append(x)
        y = x**2 - 4*x + 4
        lista_y.append(y)
    print(f'Esses são os valores de x: {lista_x}')
    print(f'Esses são os valores de x: {lista_y}')

    plt.plot(lista_x, lista_y, marker='o', linestyle='-', color='b')
    plt.title('Gráfico de uma Função Quadrática')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.grid(True)
    plt.show()

quadratica()
    
```

### Exercício 3: Gráfico de Uma Função Senoidal

Crie um gráfico da função y = sin(x) para valores de x no intervalo de 0 a 2π.

```python
import matplotlib.pyplot as plt
import numpy as np

def senoidal():
    lista_x = np.linspace(0, 2*np.pi, 100)
    lista_y = np.sin(lista_x)

    plt.plot(lista_x, lista_y)
    plt.xlabel('x (radianos)')
    plt.ylabel('sen(x)')
    plt.title('Gráfico de uma Função Senoidal de 0 a 2π')
    plt.grid(True)
    plt.show()

senoidal()
```

### Exercício 4: Gráfico de Uma Função Exponencial

Crie um gráfico para a função y = e^x para valores de x no intervalo de -2 a 2.

```python
import matplotlib.pyplot as plt
import numpy as np

def exponencial():
    x = np.linspace(-2, 2, 200)
    y = np.exp(x)

    plt.plot(x, y, color='blue', linewidth=2, label='y = e^x')
    plt.title('Gráfico da Função Exponencial y = e^x')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.grid(True)
    plt.axhline(0, color='black', linewidth=0.8) # eixo x
    plt.axvline(0, color='black', linewidth=0.8) #eixo y
    plt.legend()
    plt.show()

exponencial()
```
